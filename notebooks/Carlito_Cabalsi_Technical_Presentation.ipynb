# 2. Business Problem

Student dropout remains a major challenge for educational institutions.

Students experiencing:
- High burnout
- High stress
- Low motivation
- Low attendance

are more likely to be classified as high dropout risk.

The goal is to identify at-risk students early so intervention programs can be provided.

# 3. Dataset Overview

### Dataset Properties

- 800 student records
- 26 variables
- Binary classification target

### Target Variable

Dropout_Risk

- 0 = Low Risk
- 1 = High Risk

# 4. Data Cleaning & Preprocessing

### Activities Performed

- Removed Student_ID
- Converted target labels
- Missing value treatment
- Categorical encoding
- Feature preparation

# 5. Exploratory Data Analysis

### Key Findings

- Burnout Level showed the strongest relationship with dropout risk.
- Students with lower attendance were more likely to be classified as high risk.
- Higher motivation was associated with lower dropout risk.
- Stress and anxiety were positively related to dropout risk.
- Students with counseling access generally showed lower dropout risk.





## Burnout Level vs Dropout Risk

### Observation

- Low Burnout → 23.8% High Risk
- Medium Burnout → 63.2% High Risk
- High Burnout → 92.0% High Risk

Burnout was identified as the strongest individual predictor of dropout risk during exploratory analysis.


# 6. Feature Engineering

## Why Create New Features?

Feature engineering was performed to capture broader student characteristics that may not be fully represented by individual variables.

### Academic Engagement Index

Combined:

- Attendance Percentage
- Study Hours Per Day
- Previous GPA

Purpose:
- Represent overall academic engagement and performance.

### Mental Health Index

Combined:

- Stress Level
- Anxiety Score
- Financial Stress Score

Purpose:
- Represent overall psychological and financial burden.

## Results

Both engineered features became some of the strongest predictors during feature selection.

### Key Outcome

- Academic Engagement Index became the strongest feature in correlation analysis.
- Mental Health Index ranked among the most predictive variables.
- Both engineered features outperformed many original variables.

# 7. Feature Selection and PCA

## Feature Selection Methods

- **Filter method:** Correlation-based feature ranking
- **Embedded method:** Random Forest feature importance

## Top Predictive Features

1. Burnout Level
2. Academic Engagement Index
3. Attendance Percentage
4. Mental Health Index
5. Motivation Score

## PCA Findings

- 17 components were needed to retain 80% of the variance.
- 20 components were needed to retain 90%.
- Information was distributed across many variables.
- Original features were retained for better interpretability.



### Technical Interpretation

Random Forest confirmed the main patterns observed during EDA. Burnout Level was the strongest predictor, while both engineered indices ranked among the most useful features.

PCA did not provide a large reduction in dimensionality, so the original features were retained for model development and interpretation.

# 8. Model Implementation

## Models Evaluated

- Logistic Regression
- Decision Tree
- Random Forest
- XGBoost

## Objective

Compare the performance of different machine learning approaches and identify the model that provides the most accurate dropout risk predictions.

## Evaluation Metrics

- Accuracy
- Precision
- Recall
- F1 Score

## Model Comparison Results

| Model | Accuracy | Precision | Recall | F1 Score |
|---|---:|---:|---:|---:|
| **Logistic Regression** | **70.00%** | **72.41%** | **72.41%** | **72.41%** |
| Decision Tree | 59.38% | 61.96% | 65.52% | 63.69% |
| Random Forest | 68.13% | 70.45% | 71.26% | 70.86% |
| XGBoost | 68.13% | 70.00% | 72.41% | 71.19% |

### Key Finding

Logistic Regression achieved the highest overall performance across accuracy, precision, and F1 score, while tying with XGBoost for the top recall at 72.41%. It is selected as the best-performing baseline model, with XGBoost serving as the leading ensemble alternative.

However, across all possible decision thresholds, Random Forest achieved the highest ROC-AUC (0.778) compared to Logistic Regression (0.770) and XGBoost (0.756). This indicates that Random Forest has the strongest underlying discriminative and probability-ranking capability. If the institution tunes the decision threshold below 0.50 to prioritize identifying vulnerable students (Recall), Random Forest provides the best foundation.


# 9. Explainability & Fairness Analysis

## Model Explainability

Random Forest feature importance was used to determine which factors contributed most strongly to dropout risk predictions.

### Top Predictors

1. Burnout Level
2. Academic Engagement Index
3. Attendance Percentage
4. Mental Health Index
5. Motivation Score

The results were consistent across EDA, correlation analysis, and embedded feature selection.

## Fairness Analysis

The model was reviewed across several student groups:

- Gender
- Age Group
- Family Income Bracket
- Counseling Access

### Findings

- Gender groups showed similar dropout-risk distributions.
- Age groups showed minimal differences.
- Lower-income students showed higher dropout risk.
- Students with counseling access showed lower dropout risk.

### Responsible Use

The model should be used to support student intervention programs and not as the sole basis for decisions affecting individual students.

### SHAP Explainability & Model Interpretation Analysis

To interpret the model's nonlinear decisions, SHAP was applied using TreeExplainer on the fitted XGBoost model, which was specifically chosen to capture complex interactions and threshold effects that linear models miss.

⚬	Burnout Level, Attendance Percent, and Motivation Score are the primary predictors of student retention.

⚬ High stress, anxiety, screen time, and backlogs increase dropout risk, whereas strong attendance, motivation, and GPA act as protective buffers.

⚬ The waterfall plot demonstrated how severe burnout and low attendance drove a high-risk prediction for an individual student despite zero active backlogs.

⚬ Demographic features (Gender, Residence Type, Income Bracket) showed minimal SHAP impact, confirming that predictions are driven by actionable academic and behavioral indicators rather than demographic bias


# 10. Conclusions & Future Improvements

## Conclusions

- Burnout Level, Attendance Percent and Motivation Score emerged as the strongest predictors of dropout risk.
- Attendance, motivation, and mental health factors were strongly associated with student risk.
- Two engineered features improved the analysis:

## Future Improvements

- Collect larger, multi-institutional student datasets to enhance demographic and institutional representation.
- Integrate longitudinal tracking (e.g., semester-over-semester GPA trends, weekly LMS activity logs, and midterm exam trajectories) rather than static snapshot indicators.
- Conduct systematic hyperparameter optimization (GridSearchCV or Bayesian Optimization) for ensemble methods (XGBoost, Random Forest) to push model recall beyond the 80% benchmark.
- Deploy an early-warning intervention dashboard with role-based access for academic advisors and counseling staff.
- Establish an automated feedback loop to track the success rate of proactive interventions and continuously retrain models on newly graduated cohorts.
